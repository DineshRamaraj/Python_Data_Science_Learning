# Slicing

* If we want to get the subset of the complete part/set/group of elements then we call it slicing.
* [start : end : step]
* Start will be index value from where you start the slicing
* Stop will be +1 to the index where you want to end the slicing
* Step will be number of times you will take a jumb



```python
li = ['a', 'b', 1, 2, 3, 4, 5, 7.8, 8, 9]
print(li[::2])
print(li[3:-3:2])
print(li[:])
```

# Conditional Statements

* To make decision and execute particular section of code
* Keywords: if, else, elif

```python
# if: when we have only one condition to check (we can use if)
# elif: if we have multiple condition to check...
# else: If my other condition are not True (else block will execute)
age = 21
if age >= 18 and age <= 19:
  print("you are an adult")
  print("Sure! Please vote...")
elif age == 20:
  print("Please take responsibility of voting!")
elif age > 20:
  print("Please ask everyone else also to vote!")
else:
  print("you are an child")
  print("Please wait some year...")
```

# Nested Condition (ladder condition)

* Inside if statement we have another if statement....

```python
# Nested Condition
marks = int(input("Please enter you marks: "))
if marks >= 33:
  print("passed")
  if marks >= 85 and marks <= 100:
    print('grade A')
  elif marks >= 60 and marks < 85:
    print('grade B')
  else:
    print('Grade C')
else:
  print("Failed, Please try again")
```

# Control Flow Statement

* Break
* Continue
* Pass

```python
# Break
for i in range(1, 11):
  print(i, end=" ")
  if i == 6:
    break

print()

for i in range(1, 11):
  if i == 6:
    break
  print(i, end=" ")
```

```python
# Continue: it skips the particular or current iteration only...
for i in range(1, 11):
  if i == 6 or i == 9:
    continue
  print(i, end=' ')
```

```python
# pass: if you have not decided what to do at a certain iteration, then we use pass statement because, we cannnot keep the scope empty....
for i in range(1, 11):
  if i == 6:
    pass
  print(i)
```
