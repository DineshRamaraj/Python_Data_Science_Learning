# **List**

  1. A data type that stores multiple values/elements of any data.
  2. It follows the order of indexing.
  3. List is mutable. (change the element/update)
  4. To create a list [] square brackets.
  5. Duplicates are allowed.


```python
list1 = ['a', 'b', 'c', 1, 2, 3, True, False, 5.93, 2+5j]
print(list1)
print(type(list1))
print(list1[0])
print(type(list1[0]))
print(list1[3])
print(type(list1[3]))
print(list1[6])
print(type(list1[6]))
print(list1[8])
print(type(list1[8]))
print(list1[9])
print(type(list1[9]))
```
